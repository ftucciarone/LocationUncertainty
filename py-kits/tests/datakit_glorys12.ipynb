# Opening GLORYS12 with datakit

Monthly data, 1/12 degree. SSH has 12 snapshots; the 3D fields have 120. This grid is heavy, so the 3D examples use a box.

```python
import os, sys, time
sys.path.insert(0, os.path.abspath(".."))
from datakit import glorys12 as data

BOX = (500, 600, 300, 450)   # (j0, j1, i0, i1)

def opened(*args, **kwargs):
    """Read with datakit, then print the shape, size and time taken."""
    t0 = time.time()
    a = data.read(*args, **kwargs)
    print(f"shape {a.shape}, {a.nbytes/1e9:.3f} GB, {time.time()-t0:.1f} s")
    return a
```

## SSH: 1, 10 and all snapshots

There are only 12 SSH snapshots.

```python
a = opened("sossheig", "2015-01", "2015-01")
a = opened("sossheig", "2015-01", "2015-10")
a = opened("sossheig")
```

## U at the surface, in a box: 1, 10 and all snapshots

```python
a = opened("vozocrtx", "1993-01", "1993-01", box=BOX)
a = opened("vozocrtx", "1993-01", "1993-10", box=BOX)
a = opened("vozocrtx", box=BOX)
```

## U, in a box: 1 level, 10 levels and all levels of one snapshot

`level` takes one index or `None`. For a subset of levels, read all of them and slice.

```python
a = opened("vozocrtx", "1993-01", "1993-01", level=0, box=BOX)
a = opened("vozocrtx", "1993-01", "1993-01", level=None, box=BOX)
print("first 10 levels", a[:, :10].shape)
```
