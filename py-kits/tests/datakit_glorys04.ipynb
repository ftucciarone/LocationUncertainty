# Opening GLORYS04 with datakit

Daily data, 1/4 degree, 75 levels. U and V have 731 snapshots; SSH has 364.

```python
import os, sys, time
sys.path.insert(0, os.path.abspath(".."))
from datakit import glorys04 as data

BOX = (500, 600, 300, 450)   # (j0, j1, i0, i1)

def opened(*args, **kwargs):
    """Read with datakit, then print the shape, size and time taken."""
    t0 = time.time()
    a = data.read(*args, **kwargs)
    print(f"shape {a.shape}, {a.nbytes/1e9:.3f} GB, {time.time()-t0:.1f} s")
    return a
```

## SSH: 1, 10 and all snapshots

```python
a = opened("sossheig", "2015-01-01", "2015-01-01")
a = opened("sossheig", "2015-01-01", "2015-01-10")
a = opened("sossheig")
```

## U at the surface: 1, 10 and all snapshots

```python
a = opened("vozocrtx", "2015-01-01", "2015-01-01")
a = opened("vozocrtx", "2015-01-01", "2015-01-10")
a = opened("vozocrtx")
```

## U: 1 level, 10 levels and all levels of one snapshot

`level` takes one index or `None`. For a subset of levels, read all of them and slice.

```python
a = opened("vozocrtx", "2015-01-01", "2015-01-01", level=0)
a = opened("vozocrtx", "2015-01-01", "2015-01-01", level=None)
print("first 10 levels", a[:, :10].shape)
```

## All levels of all snapshots, in a box

The full grid would need about 320 GB, so this uses a box.

```python
a = opened("vozocrtx", level=None, box=BOX)
```

```python
a = opened("vozocrtx", "2015-01-01", "2015-01-10", baroclinic=True)
```
